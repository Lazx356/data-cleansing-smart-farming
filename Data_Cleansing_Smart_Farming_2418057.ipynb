## 3. Membaca Dataset

```python
import os

# Mencari file dataset di Google Drive
nama_file = "Smart_Farming_Crop_Yield_2024.csv"

file_path = None

for root, dirs, files in os.walk('/content/drive/MyDrive'):
    if nama_file in files:
        file_path = os.path.join(root, nama_file)
        break

print("Lokasi file:", file_path)

# Membaca dataset
df = pd.read_csv(file_path)

print("Dataset berhasil dibaca!")
print("Ukuran dataset:", df.shape)
```

```python
# Menampilkan informasi dataset
df.info()
```

## 6. Mengecek Missing Value

```python
# Mengecek jumlah missing value pada setiap kolom
df.isnull().sum()
```

## 7. Mengecek Data Duplikat

```python
# Mengecek jumlah data duplikat
df.duplicated().sum()
```

## 8. Menangani Missing Value

```python
# Mengisi missing value pada kolom kategori
df['irrigation_type'] = df['irrigation_type'].fillna('Unknown')
df['crop_disease_status'] = df['crop_disease_status'].fillna('Unknown')

print("Missing value berhasil ditangani.")
```

## 9. Verifikasi Missing Value

```python
# Mengecek kembali missing value
df.isnull().sum()
```

## 10. Mengubah Tipe Data Tanggal

```python
# Mengubah kolom tanggal menjadi format datetime
df['sowing_date'] = pd.to_datetime(df['sowing_date'])
df['harvest_date'] = pd.to_datetime(df['harvest_date'])
df['timestamp'] = pd.to_datetime(df['timestamp'])

print("Tipe data tanggal berhasil diubah.")
```

## 11. Verifikasi Tipe Data

```python
# Mengecek tipe data setelah proses cleansing
df.dtypes
```

## 12. Mengecek Nilai Data Numerik

```python
# Melihat statistik data numerik
df.describe()
```

## 13. Mengecek Konsistensi Tanggal

```python
# Mengecek apakah ada tanggal panen sebelum tanggal tanam
data_tanggal_tidak_valid = df[df['harvest_date'] < df['sowing_date']]

print("Jumlah data dengan tanggal tidak valid:", len(data_tanggal_tidak_valid))
```

## 15. Menyimpan Dataset Hasil Cleansing

```python
# Menyimpan dataset hasil cleansing ke Google Drive
output_path = '/content/drive/MyDrive/2418057_Dataset/Smart_Farming_Crop_Yield_2024_Cleansed.csv'

df.to_csv(output_path, index=False)

print("Dataset hasil cleansing berhasil disimpan!")
print("Lokasi:", output_path)
```

## 16. Verifikasi Dataset Hasil Cleansing

```python
# Membaca kembali file hasil cleansing
df_clean = pd.read_csv(output_path)

print("Ukuran dataset:", df_clean.shape)
print("Total missing value:", df_clean.isnull().sum().sum())
print("Total data duplikat:", df_clean.duplicated().sum())
```
